# 🧹 Phase 2: Automated Data Quality Engine

Data quality is the foundation of reliable AI systems.

Before performing analytics or machine learning, DataPilot AI evaluates:

- Completeness
- Consistency
- Uniqueness
- Data types
- Potential issues

The output is a Dataset Health Report.

```python
# ==========================================
# DATASET SUMMARY ENGINE
# ==========================================

def dataset_summary(df):

    summary = {
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Duplicate Rows": df.duplicated().sum(),
        "Missing Cells": df.isnull().sum().sum(),
        "Memory Usage MB": round(
            df.memory_usage(deep=True).sum()/1024**2,
            2
        )
    }

    return pd.DataFrame(
        summary.items(),
        columns=["Metric","Value"]
    )


dataset_summary(df)
```

## Missing Value Analysis

Missing data can affect:

- Statistical analysis
- Machine learning models
- Business decisions

The AI agent identifies columns requiring attention.

```python
# ==========================================
# MISSING VALUE DETECTOR
# ==========================================

def missing_value_report(df):

    missing = pd.DataFrame({
        "Column": df.columns,
        "Missing Values": df.isnull().sum(),
        "Missing Percentage":
        (df.isnull().mean()*100).round(2)
    })

    missing = missing[
        missing["Missing Values"] > 0
    ]

    if len(missing)==0:
        return pd.DataFrame(
            {
                "Message":
                ["No missing values detected ✅"]
            }
        )

    return missing.sort_values(
        "Missing Percentage",
        ascending=False
    )


missing_value_report(df)
```

## Duplicate Detection

Duplicate records can inflate:

- Revenue
- Customer counts
- Model accuracy

The system detects duplicate transactions.

```python
# ==========================================
# DUPLICATE ANALYZER
# ==========================================

def duplicate_report(df):

    duplicates = df[df.duplicated()]

    result = {
        "Duplicate Rows":
        len(duplicates),

        "Duplicate Percentage":
        round(
            len(duplicates)/len(df)*100,
            2
        )
    }

    return pd.DataFrame(
        result.items(),
        columns=["Metric","Value"]
    )


duplicate_report(df)
```

## Column Type Intelligence

The AI analyst automatically separates:

- Numerical features
- Categorical features
- Date columns

```python
# ==========================================
# FEATURE TYPE DETECTOR
# ==========================================

def feature_analysis(df):

    numerical = df.select_dtypes(
        include=np.number
    ).columns.tolist()


    categorical = df.select_dtypes(
        include="object"
    ).columns.tolist()


    dates = df.select_dtypes(
        include="datetime"
    ).columns.tolist()


    return {
        "Numerical Columns": numerical,
        "Categorical Columns": categorical,
        "Date Columns": dates
    }


feature_analysis(df)
```

## Data Health Score

DataPilot AI calculates a score out of 100.

Scoring:

Completeness:
40 points

Duplicate Quality:
30 points

Data Type Consistency:
30 points

```python
# ==========================================
# DATA HEALTH SCORE
# ==========================================


def data_health_score(df):

    total_cells = df.shape[0]*df.shape[1]


    missing_ratio = (
        df.isnull().sum().sum()
        /
        total_cells
    )


    completeness_score = (
        1-missing_ratio
    )*40



    duplicate_ratio = (
        df.duplicated().sum()
        /
        len(df)
    )


    uniqueness_score = (
        1-duplicate_ratio
    )*30



    type_score = 30



    score = (
        completeness_score
        +
        uniqueness_score
        +
        type_score
    )


    return round(score,2)



health = data_health_score(df)

print(
    f"Dataset Health Score: {health}/100"
)
```

## AI Cleaning Recommendations

The system explains what should be done.

```python
# ==========================================
# CLEANING RECOMMENDATION ENGINE
# ==========================================


def generate_recommendations(df):

    recommendations=[]


    missing = df.isnull().sum().sum()

    duplicates = df.duplicated().sum()


    if missing > 0:
        recommendations.append(
            "⚠️ Missing values detected. Apply imputation or removal."
        )
    else:
        recommendations.append(
            "✅ Dataset has no missing values."
        )


    if duplicates > 0:
        recommendations.append(
            "⚠️ Duplicate transactions detected. Review duplicates."
        )
    else:
        recommendations.append(
            "✅ No duplicate records found."
        )


    if len(df.select_dtypes(
        include=np.number).columns)>0:

        recommendations.append(
            "✅ Numerical features detected for statistical analysis."
        )


    recommendations.append(
        "🤖 Dataset is ready for exploratory analysis."
    )


    return recommendations



for item in generate_recommendations(df):
    print(item)
```

# 🧽 Automated Cleaning Pipeline

The agent creates a cleaned copy.

Operations:

- Remove duplicates
- Standardize column names
- Handle missing values

```python
# ==========================================
# AUTOMATIC CLEANING
# ==========================================


def clean_dataset(df):

    clean_df = df.copy()


    # standardize columns

    clean_df.columns = (
        clean_df.columns
        .str.lower()
        .str.replace(
            " ",
            "_"
        )
    )


    # remove duplicates

    clean_df = (
        clean_df
        .drop_duplicates()
    )


    # fill numerical missing values

    numerical = clean_df.select_dtypes(
        include=np.number
    ).columns


    for col in numerical:
        clean_df[col] = (
            clean_df[col]
            .fillna(
                clean_df[col].median()
            )
        )


    # fill categorical missing values

    categorical = clean_df.select_dtypes(
        include="object"
    ).columns


    for col in categorical:

        clean_df[col] = (
            clean_df[col]
            .fillna("Unknown")
        )


    return clean_df



clean_df = clean_dataset(df)


clean_df.head()
```

#Rule-Based Intelligence Layer
Before using expensive LLMs, good AI systems use deterministic logic.

This gives:

Faster responses

Lower cost

More reliable calculations

```python

# ==========================================
# ANALYTICAL QUESTION ENGINE
# ==========================================


def data_question_agent(question, df):

    question = question.lower()


    # Highest sales region

    if "highest sales region" in question or "best region" in question:

        result = (
            df.groupby("region")
            ["total_sales"]
            .sum()
            .sort_values(
                ascending=False
            )
            .head(1)
        )

        return (
            f"🌍 The highest sales region is "
            f"{result.index[0]} "
            f"with total sales of "
            f"{result.values[0]:,.2f}"
        )



    # Best product

    elif "best product" in question or "top product" in question:

        result = (
            df.groupby("product_name")
            ["total_sales"]
            .sum()
            .sort_values(
                ascending=False
            )
            .head(1)
        )

        return (
            f"🏆 The best performing product is "
            f"{result.index[0]} "
            f"with sales of "
            f"{result.values[0]:,.2f}"
        )



    # Average satisfaction

    elif "satisfaction" in question or "csat" in question:

        score = (
            df["csat_score"]
            .mean()
        )

        return (
            f"😊 Average customer satisfaction score is "
            f"{score:.2f}"
        )



    # Churn analysis

    elif "churn" in question:

        churn = (
            df["churn_risk"]
            .value_counts()
        )

        return (
            f"⚠️ Churn Risk Distribution:\n\n{churn}"
        )



    # Profit

    elif "profit" in question:

        profit = (
            df["net_profit"]
            .sum()
        )

        return (
            f"💰 Total net profit is "
            f"{profit:,.2f}"
        )


    else:

        return (
            "🤖 I understand your question, "
            "but this analysis is not yet supported. "
            "Try asking about sales, profit, products, "
            "regions, customers, or churn."
        )
```

#Build Insight Engine
The AI will automatically discover:
Best region
Best product

Customer behaviour

```python
# ==========================================
# ADVANCED INSIGHT ENGINE
# ==========================================


def generate_business_insights(df):

    insights = []


    # Revenue leader

    if "Region" in df.columns:

        region = (
            df.groupby("region")
            ["total_sales"]
            .sum()
            .idxmax()
        )

        sales = (
            df.groupby("region")
            ["total_sales"]
            .sum()
            .max()
        )

        insights.append(
            f"🌍 {region} is the strongest region, generating sales of {sales:,.2f}."
        )


    # Product leader

    if "Product_Name" in df.columns:

        product = (
            df.groupby("product_name")
            ["total_sales"]
            .sum()
            .idxmax()
        )

        insights.append(
            f"🏆 {product} is the highest performing product."
        )


    # Profit

    if "Net_Profit" in df.columns:

        profit = df["net_profit"].sum()

        insights.append(
            f"💰 Total net profit generated is {profit:,.2f}."
        )


    # Customer satisfaction

    if "CSAT_Score" in df.columns:

        csat = df["csat_score"].mean()

        insights.append(
            f"😊 Average customer satisfaction score is {csat:.2f}/5."
        )


    # Churn

    if "Churn_Risk" in df.columns:

        churn = (
            df["churn_risk"]
            .value_counts()
            .idxmax()
        )

        insights.append(
            f"⚠️ Most common churn category is {churn}."
        )


    return insights



insights = generate_business_insights(eda_df)


for i in insights:
    print(i)
```

```python
# ==========================================
# DATA UPLOAD FUNCTION
# ==========================================


def upload_data(file):

    global APP_DATA


    path = file.name


    if path.endswith(".csv"):

        df = pd.read_csv(path)

    elif path.endswith(".xlsx"):

        df = pd.read_excel(path)

    else:

        return "Unsupported file format"



    APP_DATA["raw"] = df


    APP_DATA["clean"] = clean_dataset(df)



    message = f"""
✅ Dataset Loaded Successfully

Rows:
{df.shape[0]}

Columns:
{df.shape[1]}

Health Score:
{data_health_score(df)}/100
"""


    return message
```

Automated Dashboard Generator

```python
# ==========================================
# DASHBOARD CREATOR
# ==========================================


def create_dashboard():

    df = APP_DATA["clean"]


    if df is None:

        return None,None,None



    # Revenue by region

    region = (
        df.groupby("region")
        ["total_sales"]
        .sum()
        .reset_index()
    )


    fig1 = px.bar(
        region,
        x="region",
        y="total_sales",
        title="Revenue By Region"
    )



    # Products

    products = (
        df.groupby("product_name")
        ["total_sales"]
        .sum()
        .sort_values(
            ascending=False
        )
        .head(10)
        .reset_index()
    )


    fig2 = px.bar(
        products,
        x="total_sales",
        y="product_name",
        orientation="h",
        title="Top Products"
    )



    # Customer segments

    segments = (
        df.groupby("customer_segment")
        ["total_sales"]
        .sum()
        .reset_index()
    )


    fig3 = px.pie(
        segments,
        names="customer_segment",
        values="total_sales",
        title="Customer Segments"
    )


    return fig1,fig2,fig3
```

#Create Data Analysis Tools

```python
# =====================================
# STATISTICS TOOL
# =====================================


def statistics_tool(column):

    df = APP_DATA["clean"]


    if column not in df.columns:

        return "Column does not exist"


    return df[column].describe().to_string()
```

```python
# =====================================
# GROUP ANALYSIS TOOL
# =====================================


def group_analysis(
    category,
    metric
):

    df = APP_DATA["clean"]


    result = (
        df.groupby(category)[metric]
        .sum()
        .sort_values(
            ascending=False
        )
    )


    return result.head(10).to_string()
```

```python
# =====================================
# CORRELATION TOOL
# =====================================


def correlation_tool():

    df = APP_DATA["clean"]


    corr = (
        df.select_dtypes(
            include=np.number
        )
        .corr()
    )


    return corr.to_string()
```

```python
%%writefile DataPilot-AI-Agent/app.py

import gradio as gr
import pandas as pd
import numpy as np
import plotly.express as px
from fpdf import FPDF


DATA = {
    "raw": None,
    "clean": None
}


def clean_dataset(df):

    df = df.copy()

    df.columns = (
        df.columns
        .str.lower()
        .str.replace(" ","_")
    )


    df = df.drop_duplicates()


    for col in df.select_dtypes(
        include=np.number
    ):

        df[col] = df[col].fillna(
            df[col].median()
        )


    for col in df.select_dtypes(
        include="object"
    ):

        df[col] = df[col].fillna(
            "Unknown"
        )


    return df



def upload_data(file):

    if file.name.endswith(".csv"):

        df = pd.read_csv(
            file.name
        )

    else:

        df = pd.read_excel(
            file.name
        )


    DATA["raw"] = df

    DATA["clean"] = clean_dataset(df)


    return f"""
Dataset loaded successfully

Rows:
{df.shape[0]}

Columns:
{df.shape[1]}
"""



def preview():

    return DATA["clean"].head(20)



def dashboard():

    df = DATA["clean"]


    region = (
        df.groupby("region")
        ["total_sales"]
        .sum()
        .reset_index()
    )


    fig1 = px.bar(
        region,
        x="region",
        y="total_sales",
        title="Sales By Region"
    )


    product = (
        df.groupby("product_name")
        ["total_sales"]
        .sum()
        .sort_values(
            ascending=False
        )
        .head(10)
        .reset_index()
    )


    fig2 = px.bar(
        product,
        x="total_sales",
        y="product_name",
        orientation="h",
        title="Top Products"
    )


    return fig1,fig2



def analyst(question):

    df = DATA["clean"]


    q = question.lower()


    if "profit" in q:

        return (
            f"Total profit is "
            f"{df['net_profit'].sum():,.2f}"
        )


    if "region" in q:

        result = (
            df.groupby("region")
            ["total_sales"]
            .sum()
            .idxmax()
        )


        return (
            f"Highest sales region: {result}"
        )


    if "product" in q:

        result = (
            df.groupby("product_name")
            ["total_sales"]
            .sum()
            .idxmax()
        )


        return (
            f"Best product: {result}"
        )


    return (
        "Ask about sales, profit, regions or products."
    )



with gr.Blocks(
    title="DataPilot AI"
) as demo:


    gr.Markdown(
    """
    # 🤖 DataPilot AI

    Your Intelligent Data Scientist
    """
    )


    with gr.Tab("Upload"):

        file = gr.File()

        status = gr.Textbox()


        file.upload(
            upload_data,
            file,
            status
        )



    with gr.Tab("Preview"):

        button = gr.Button(
            "Show Data"
        )

        table = gr.Dataframe()


        button.click(
            preview,
            outputs=table
        )



    with gr.Tab("Dashboard"):

        button = gr.Button(
            "Generate Dashboard"
        )

        chart1 = gr.Plot()

        chart2 = gr.Plot()


        button.click(
            dashboard,
            outputs=[
                chart1,
                chart2
            ]
        )



    with gr.Tab("AI Analyst"):

        question = gr.Textbox()

        answer = gr.Textbox()


        ask = gr.Button(
            "Ask DataPilot"
        )


        ask.click(
            analyst,
            question,
            answer
        )



demo.launch()
```
